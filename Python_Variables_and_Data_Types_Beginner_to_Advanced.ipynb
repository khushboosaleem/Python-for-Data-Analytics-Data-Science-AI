# 1. What is a Variable?

A **variable** is a name used to store a value in memory.

In Python, you do not need to declare the type of a variable before using it.

````python
name = "sana"
age = 20
salary = 50000

print(name)
print(age)
print(salary)
````

Python automatically detects the type of value stored in a variable.  
This is called **dynamic typing**.

# 2. Creating and Updating Variables

````python
score = 50
print("Initial score:", score)

score = 75
print("Updated score:", score)

score = score + 10
print("Final score:", score)
````

# 3. Variable Naming Rules

### Valid Rules
- A variable name can contain letters, numbers, and underscores.
- It cannot start with a number.
- It cannot contain spaces.
- Python keywords cannot be used as variable names.
- Python variable names are case sensitive.

### Recommended Style
Use **snake_case** for variable names.

````python
student_name = "Ayesha"
total_marks = 950
is_passed = True

print(student_name)
print(total_marks)
print(is_passed)
````

### Invalid Examples

```python
2name = "sana"        # starts with number
student name = "sana" # contains space
class = "Python"     # class is a Python keyword
```

# 7. Integer (`int`)

Integers are whole numbers without decimal points.

````python
positive_number = 100
negative_number = -25
zero = 0

print(type(positive_number))
print(type(negative_number))
print(type(zero))
````

Python integers can be very large because Python automatically manages integer size.

````python
big_number = 123456789012345678901234567890
print(big_number)
print(type(big_number))
````

# 8. Float (`float`)

Floats contain decimal values.

````python
height = 5.8
temperature = -2.5
price = 499.99

print(height)
print(type(height))
````

Scientific notation is also supported.

````python
small_number = 1.5e-4
large_number = 2.5e6

print(small_number)
print(large_number)
````

# 9. Complex Numbers (`complex`)

# 10. String (`str`)

# 11. Boolean (`bool`)

Boolean values are either `True` or `False`.

````python
is_student = True
is_logged_in = False

print(type(is_student))
print(type(is_logged_in))
````

````python
age = 20

print(age >= 18)
print(age < 18)
print(10 == 10)
print(10 != 5)
````

# 12. List (`list`)

A list:
- Stores multiple values
- Is ordered
- Is mutable
- Allows duplicate values

````python
students = ["Ali", "Sara", "Ahmed"]

print(students)
print(type(students))
print(students[0])
````

# 13. Tuple (`tuple`)

A tuple is:
- Ordered
- Immutable
- Allows duplicate values

# 14. Set (`set`)

A set:

Python allows variable swapping without a temporary variable.

# 23. Unpacking Collections

````python
student = ("Ali", 20, "Python")

name, age, course = student

print(name)
print(age)
print(course)
````

## Extended Unpacking

````python
numbers = [1, 2, 3, 4, 5, 6]

first, second, *remaining = numbers

print(first)
print(second)
print(remaining)
````

````python
first, *middle, last = [10, 20, 30, 40, 50]

print(first)
print(middle)
print(last)
````

# 24. Dynamic Typing

Python variables can hold different types at different times.

````python
value = 10
print(value, type(value))

value = "Python"
print(value, type(value))

value = 10.5
print(value, type(value))
````

This flexibility is useful, but changing a variable's type unnecessarily can make code harder to understand.

# 25. Mutable vs Immutable Data Types

## Immutable
These cannot be changed after creation:
- `int`
- `float`
- `bool`
- `str`
- `tuple`
- `frozenset`
- `bytes`

## Mutable
These can be changed:
- `list`
- `dict`
- `set`
- `bytearray`

## Mutable Example

````python
numbers = [1, 2, 3]

print("Before:", numbers)

numbers.append(4)

print("After:", numbers)
````

## Immutable Example

````python
name = "Ali"

print(name)

# Strings cannot be changed directly:
# name[0] = "S"   # TypeError

name = "S" + name[1:]
print(name)
````

# 26. Object Identity with `id()`

Python variables reference objects in memory.

`id()` returns an identifier for an object during its lifetime.

````python
x = 100
y = x

print(id(x))
print(id(y))
print(x is y)
````

# 27. `==` vs `is`

- `==` checks whether values are equal.
- `is` checks whether two variables refer to the same object.

````python
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)
print(a is b)
````

Use `is` mainly for identity checks such as:

```python
value is None
```

````python
value = None

print(value is None)
````

# 28. Reference Behavior with Mutable Objects

Two variables can refer to the same mutable object.

````python
list1 = [1, 2, 3]
list2 = list1

list2.append(4)

print("list1:", list1)
print("list2:", list2)
````

If you want an independent list, make a copy.

````python
list1 = [1, 2, 3]
list2 = list1.copy()

list2.append(4)

print("list1:", list1)
print("list2:", list2)
````

# 29. Shallow Copy vs Deep Copy

````python
import copy

original = [[1, 2], [3, 4]]

shallow = copy.copy(original)
deep = copy.deepcopy(original)

original[0][0] = 999

print("Original:", original)
print("Shallow:", shallow)
print("Deep:", deep)
````

A **shallow copy** copies the outer container but nested objects may still be shared.

A **deep copy** recursively copies nested objects.

# 30. Type Checking with `isinstance()`

`isinstance()` is often better than using `type()` for type checking.

````python
age = 20

print(isinstance(age, int))
print(isinstance(age, str))
````

````python
value = 10.5

print(isinstance(value, (int, float)))
````

# 31. Type Hints

Python supports optional type hints to improve readability and help IDEs/type checkers.

````python
name: str = "Ali"
age: int = 20
height: float = 5.9
is_student: bool = True

print(name, age, height, is_student)
````

Type hints do not normally enforce types at runtime.

````python
value: int = 10
print(value)

value = "Python"
print(value)
````

# 32. Type Hints in Functions

````python
def add_numbers(a: int, b: int) -> int:
    return a + b

print(add_numbers(10, 20))
````

# 33. Advanced Type Hints

````python
from typing import Optional

student_name: str = "Ali"
marks: list[int] = [80, 90, 85]
student: dict[str, int] = {"age": 20, "marks": 90}
phone_number: Optional[str] = None

print(student_name)
print(marks)
print(student)
print(phone_number)
````

In modern Python, union types can also be written using `|`.

````python
value: int | str = 10
print(value)

value = "Python"
print(value)
````

# 34. Constants in Python

Python does not enforce true constants, but programmers use uppercase names by convention.

````python
PI = 3.14159
MAX_USERS = 100

print(PI)
print(MAX_USERS)
````

# 35. Numeric Precision

Floating-point numbers may not always represent decimal values exactly.

````python
print(0.1 + 0.2)
print((0.1 + 0.2) == 0.3)
````

For financial calculations where exact decimal arithmetic is important, use `Decimal`.

````python
from decimal import Decimal

a = Decimal("0.1")
b = Decimal("0.2")

print(a + b)
print((a + b) == Decimal("0.3"))
````

# 37. `memoryview`

`memoryview` allows efficient access to binary data without copying it.

````python
data = bytearray(b"Python")
view = memoryview(data)

print(view)
print(view[0])
````

# 38. Variable Scope Introduction

Variables can exist in different scopes.

- Local
- Global
- Enclosing
- Built in

````python
global_variable = "I am global"

def demo():
    local_variable = "I am local"
    print(local_variable)
    print(global_variable)

demo()
print(global_variable)
````

# 39. Global Keyword

````python
count = 0

def update_count():
    global count
    count = 10

update_count()
print(count)
````

Use global variables carefully because too many global variables can make programs difficult to maintain.

# 40. The Walrus Operator `:=`

The walrus operator assigns a value as part of an expression.

````python
numbers = [1, 2, 3, 4, 5]

if (length := len(numbers)) > 3:
    print("The list contains", length, "items.")
````
