## What is a Function?

A **function** is a reusable block of code that performs a specific task.

Instead of writing the same code repeatedly, we write it once inside a function and call it whenever needed.

### Simple idea

**Input → Function Processing → Output**

Functions help us follow the **DRY principle**:

> **Don't Repeat Yourself**

## Why Do We Need Functions in Python?

Functions help us:

1. **Reuse code**  write once and use many times.
2. **Reduce repetition**  avoid duplicate code.
3. **Organize programs**  divide a large program into smaller parts.
4. **Improve readability**  meaningful function names explain the purpose.
5. **Make debugging easier** errors can be checked in one specific block.
6. **Maintain code easily** update the function once and all calls use the new logic.
7. **Support teamwork** different people can work on different functions.

## 1. Creating and Calling a Simple Function

We create a function with the `def` keyword.


To run it:

````python
def greet():
    print("Welcome to Python Functions!")

greet()
````

## 2. Function With a Parameter

A **parameter** is a variable written inside the function definition.

An **argument** is the actual value passed when calling the function.

````python
def greet_user(name):
    print(f"Welcome, {name}!")

greet_user("Khushboo")
greet_user("Sara")
````

## 3. Function With Multiple Parameters

````python
def introduce(name, profession):
    print(f"My name is {name}, and I am a {profession}.")

introduce("Khushboo", "Data Science Trainer")
````

## 4. Returning a Value

The `return` keyword sends a result back from the function.

A returned value can be saved in a variable and used later.

````python
def add_numbers(a, b):
    return a + b

result = add_numbers(10, 5)
print("Result:", result)
````

## `print()` vs `return`

| `print()` | `return` |
|---|---|
| Displays a value on the screen | Sends a value back to the caller |
| Mainly used to show information | Used when the result is needed later |
| Cannot directly support further calculations | Can be saved and reused |

````python
def add_with_print(a, b):
    print(a + b)

def add_with_return(a, b):
    return a + b

printed_result = add_with_print(4, 6)
returned_result = add_with_return(4, 6)

print("Value stored from print function:", printed_result)
print("Value stored from return function:", returned_result)
print("Returned value multiplied by 2:", returned_result * 2)
````

## 5. Default Arguments

A default argument is used when the caller does not provide a value.

````python
def welcome(name, course="Python"):
    print(f"Welcome {name}! You are learning {course}.")

welcome("Ali")
welcome("Sara", "Data Science")
````

## 6. Keyword Arguments

Keyword arguments allow us to specify parameter names while calling a function.

````python
def student_info(name, age, city):
    print(f"Name: {name}, Age: {age}, City: {city}")

student_info(city="Karachi", name="Ayesha", age=23)
````

## 7. Types of Functions in Python

### Common categories

1. **Built-in functions** already provided by Python  
   Examples: `print()`, `len()`, `sum()`, `max()`

2. **User defined functions** created by the programmer using `def`

3. **Anonymous or lambda functions** short functions written with `lambda`

4. **Recursive functions** functions that call themselves

````python
numbers = [10, 20, 30, 40]

print("Length:", len(numbers))
print("Total:", sum(numbers))
print("Maximum:", max(numbers))
````

## 8. Function With a Condition

````python
def check_even_odd(number):
    if number % 2 == 0:
        return "Even"
    return "Odd"

print(check_even_odd(12))
print(check_even_odd(7))
````

## 9. Function With a Loop

````python
def calculate_squares(numbers):
    squares = []

    for number in numbers:
        squares.append(number ** 2)

    return squares

values = [1, 2, 3, 4, 5]
print(calculate_squares(values))
````

## 10. Local and Global Variables

- A **local variable** is created inside a function and normally works only there.
- A **global variable** is created outside functions and can be accessed throughout the program.

````python
course = "Python"  # Global variable

def show_course():
    trainer = "Khushboo"  # Local variable
    print("Course:", course)
    print("Trainer:", trainer)

show_course()
print("Global variable outside function:", course)

# print(trainer)  # This would cause a NameError
````

## 11. Using `*args`

`*args` allows a function to receive any number of positional arguments.

Inside the function, `args` behaves like a tuple.

````python
def add_all_numbers(*numbers):
    return sum(numbers)

print(add_all_numbers(10, 20))
print(add_all_numbers(1, 2, 3, 4, 5))
````

## 12. Using `**kwargs`

`**kwargs` allows a function to receive any number of keyword arguments.

Inside the function, `kwargs` behaves like a dictionary.

````python
def display_profile(**details):
    for key, value in details.items():
        print(f"{key.title()}: {value}")

display_profile(
    name="Khushboo",
    profession="Trainer",
    skill="Data Science"
)
````

## 13. Lambda Functions

A lambda function is a small anonymous function, usually written in one line.

### Syntax

```python
lambda arguments: expression
```

Use lambda for short and simple operations. For complex logic, use a normal `def` function.

````python
square = lambda number: number ** 2
add = lambda a, b: a + b

print(square(6))
print(add(4, 8))
````

## 14. Recursive Function

A recursive function calls itself.

Every recursive function needs a **base case** to stop the recursion.

````python
def factorial(number):
    if number == 0 or number == 1:
        return 1

    return number * factorial(number - 1)

print("Factorial of 5:", factorial(5))
````

# Practical Examples

## Example 1: Calculate Discounted Price

````python
def calculate_discount(price, discount_percentage):
    discount_amount = price * discount_percentage / 100
    final_price = price - discount_amount
    return final_price

print("Final price:", calculate_discount(5000, 10))
````

## Example 2: Student Grade Calculator

````python
def calculate_grade(marks):
    if marks >= 80:
        return "A"
    elif marks >= 70:
        return "B"
    elif marks >= 60:
        return "C"
    elif marks >= 50:
        return "D"
    return "Fail"

student_marks = 76
print("Grade:", calculate_grade(student_marks))
````

## Example 3: Clean a List of Names

````python
def clean_names(names):
    return [name.strip().title() for name in names]

raw_names = [" khushboo", "SARA ", " ali "]
print(clean_names(raw_names))
````

## Example 4: Data Analysis Helper Function

````python
def summarize_numbers(numbers):
    if not numbers:
        return {
            "count": 0,
            "total": 0,
            "average": None,
            "minimum": None,
            "maximum": None
        }

    return {
        "count": len(numbers),
        "total": sum(numbers),
        "average": sum(numbers) / len(numbers),
        "minimum": min(numbers),
        "maximum": max(numbers)
    }

sales = [1200, 1500, 900, 1800, 1600]
summary = summarize_numbers(sales)

for key, value in summary.items():
    print(f"{key.title()}: {value}")
````

# Best Practices

- Give functions meaningful names.
- Keep each function focused on one task.
- Use `return` when the result will be reused.
- Avoid writing very long functions.
- Add a docstring to explain important functions.
- Do not use global variables unnecessarily.
- Use lambda only for small and simple expressions.
- Test functions with different inputs.

## Function With a Docstring

````python
def calculate_average(numbers):
    """Return the average of a non empty list of numbers."""

    if not numbers:
        raise ValueError("The numbers list cannot be empty.")

    return sum(numbers) / len(numbers)

print(calculate_average([10, 20, 30]))
print(calculate_average.__doc__)
````
