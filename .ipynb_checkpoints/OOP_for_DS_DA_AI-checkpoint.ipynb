# 2. Basic Class

A class is a blueprint used to create objects.

```python
# Create a class named Student.
class Student:
    # Create a method that displays a message.
    def learn(self):
        # Print a learning message.
        print("The student is learning Python.")

# Create an object from the Student class.
student_one = Student()

# Call the method using the object.
student_one.learn()
```

# 3. Objects

An object is an instance created from a class.

```python
# Create a class named Laptop.
class Laptop:
    # Create a method that displays laptop information.
    def show_info(self):
        # Print the laptop brand.
        print("Brand:", self.brand)
        # Print the laptop RAM.
        print("RAM:", self.ram)

# Create the first Laptop object.
laptop_one = Laptop()

# Store the brand in the object.
laptop_one.brand = "Dell"

# Store the RAM in the object.
laptop_one.ram = "16 GB"

# Display the object information.
laptop_one.show_info()
```

# 4. Constructor and self

`__init__()` runs automatically when an object is created. `self` represents the current object.

```python
# Create a class named Employee.
class Employee:
    # Define a constructor that receives employee data.
    def __init__(self, name, salary):
        # Store the employee name in the current object.
        self.name = name
        # Store the employee salary in the current object.
        self.salary = salary

    # Create a method that displays employee data.
    def show_details(self):
        # Print the employee name.
        print("Name:", self.name)
        # Print the employee salary.
        print("Salary:", self.salary)

# Create an Employee object with constructor values.
employee_one = Employee("Ali", 70000)

# Display employee details.
employee_one.show_details()
```

# 5. Instance Attributes and Class Attributes

Instance attributes belong to individual objects. Class attributes are shared by all objects.

```python
# Create a class named DataCourse.
class DataCourse:
    # Create a class attribute shared by all objects.
    institute = "Data Learning Academy"

    # Define a constructor for course data.
    def __init__(self, course_name, duration):
        # Store the course name in the object.
        self.course_name = course_name
        # Store the course duration in the object.
        self.duration = duration

# Create the first course object.
course_one = DataCourse("Data Science", "6 Months")

# Create the second course object.
course_two = DataCourse("Data Analytics", "4 Months")

# Display the shared class attribute.
print(DataCourse.institute)

# Display the first object's course name.
print(course_one.course_name)

# Display the second object's course name.
print(course_two.course_name)
```

# 6. Instance Methods

Instance methods work with object-specific data.

```python
# Create a class named MobilePhone.
class MobilePhone:
    # Define a constructor for mobile phone data.
    def __init__(self, brand, price):
        # Store the mobile phone brand.
        self.brand = brand
        # Store the mobile phone price.
        self.price = price

    # Create a method that displays mobile phone information.
    def show_details(self):
        # Print the mobile phone brand.
        print("Brand:", self.brand)
        # Print the mobile phone price.
        print("Price:", self.price)

# Create a MobilePhone object.
phone = MobilePhone("Samsung", 85000)

# Display the mobile phone details.
phone.show_details()
```

# 7. Encapsulation

Encapsulation hides internal data and protects it from direct access.

```python
# Create a class named BankAccount.
class BankAccount:
    # Define a constructor for account data.
    def __init__(self, owner, balance):
        # Store the account owner's name.
        self.owner = owner
        # Store the balance as a private attribute.
        self.__balance = balance

    # Create a method for depositing money.
    def deposit(self, amount):
        # Check that the amount is positive.
        if amount > 0:
            # Add the amount to the private balance.
            self.__balance += amount

    # Create a method for viewing the balance.
    def show_balance(self):
        # Print the private balance safely.
        print("Current Balance:", self.__balance)

# Create a BankAccount object.
account = BankAccount("Sara", 5000)

# Deposit money into the account.
account.deposit(2000)

# Display the updated balance.
account.show_balance()
```

# 8. Inheritance

Inheritance allows a child class to reuse and extend the parent class.

```python
# Create a parent class named Person.
class Person:
    # Create a method that displays a person message.
    def show_person(self):
        # Print the parent class message.
        print("This is a person.")

# Create a child class named Student.
class Student(Person):
    # Create a method that displays a student message.
    def study(self):
        # Print the child class message.
        print("The student is studying.")

# Create a Student object.
student = Student()

# Call the inherited parent method.
student.show_person()

# Call the child class method.
student.study()
```

# 9. Multilevel Inheritance

A class can inherit from another child class.

```python
# Create a base class named DataLoader.
class DataLoader:
    # Create a data loading method.
    def load(self):
        # Print a loading message.
        print("Loading data...")

# Create a child class named Cleaner.
class Cleaner(DataLoader):
    # Create a data cleaning method.
    def clean(self):
        # Print a cleaning message.
        print("Cleaning data...")

# Create another child class named Trainer.
class Trainer(Cleaner):
    # Create a model training method.
    def train(self):
        # Print a training message.
        print("Training model...")

# Create a Trainer object.
pipeline = Trainer()

# Call the inherited loading method.
pipeline.load()

# Call the inherited cleaning method.
pipeline.clean()

# Call the training method.
pipeline.train()
```

# 10. Multiple Inheritance

A class can inherit features from more than one parent class.

```python
# Create a class that provides logging features.
class Logger:
    # Create a logging method.
    def log(self, message):
        # Print the log message.
        print("LOG:", message)

# Create a class that provides validation features.
class Validator:
    # Create a validation method.
    def validate(self, value):
        # Return True when the value is not missing.
        return value is not None

# Create a class that inherits from both parent classes.
class DataPipeline(Logger, Validator):
    # Create a processing method.
    def process(self, value):
        # Check whether the value is valid.
        if self.validate(value):
            # Log a success message.
            self.log("Value processed successfully.")

# Create a DataPipeline object.
pipeline = DataPipeline()

# Process a sample value.
pipeline.process(25)
```

# 11. Method Overriding

A child class can replace a parent class method with its own version.

```python
# Create a parent class named Vehicle.
class Vehicle:
    # Create a method that displays a general movement message.
    def move(self):
        # Print the parent class message.
        print("The vehicle is moving.")

# Create a child class named Car.
class Car(Vehicle):
    # Override the parent method.
    def move(self):
        # Print the child class message.
        print("The car is driving on the road.")

# Create a Car object.
car = Car()

# Call the overridden method.
car.move()
```

# 12. Polymorphism

The same method name can behave differently for different objects.

```python
# Create a class named Cat.
class Cat:
    # Create a sound method for a cat.
    def sound(self):
        # Print the cat sound.
        print("Meow")

# Create a class named Dog.
class Dog:
    # Create a sound method for a dog.
    def sound(self):
        # Print the dog sound.
        print("Woof")

# Create different animal objects.
animals = [Cat(), Dog()]

# Loop through every animal object.
for animal in animals:
    # Call the same method on different objects.
    animal.sound()
```

# 13. Magic Methods

Magic methods customize object behavior.

```python
# Create a class named Report.
class Report:
    # Define a constructor for report information.
    def __init__(self, title, pages):
        # Store the report title.
        self.title = title
        # Store the number of pages.
        self.pages = pages

    # Customize the printable object representation.
    def __str__(self):
        # Return formatted report information.
        return f"{self.title} contains {self.pages} pages."

    # Customize the length behavior.
    def __len__(self):
        # Return the number of pages.
        return self.pages

# Create a Report object.
report = Report("Sales Analysis", 15)

# Print the object using the custom string method.
print(report)

# Display the custom object length.
print(len(report))
```

# 14. Composition

Composition means one class contains an object of another class.

```python
# Create a class named Engine.
class Engine:
    # Create a method that starts the engine.
    def start(self):
        # Print the engine message.
        print("Engine started.")

# Create a class named Car.
class Car:
    # Define a constructor for the car.
    def __init__(self):
        # Create an Engine object inside the Car object.
        self.engine = Engine()

    # Create a method that starts the car.
    def start_car(self):
        # Start the internal engine object.
        self.engine.start()
        # Print the car message.
        print("Car is ready to drive.")

# Create a Car object.
car = Car()

# Start the car.
car.start_car()
```

# 15. OOP in Data Analytics Solved Mini Project

This project creates a reusable sales analysis class.

```python
# Import pandas for data analysis.
import pandas as pd

# Create a class named SalesAnalyzer.
class SalesAnalyzer:
    # Define a constructor that receives sales data.
    def __init__(self, data):
        # Store the provided DataFrame.
        self.data = data

    # Create a method that calculates total sales.
    def total_sales(self):
        # Return the sum of the Sales column.
        return self.data["Sales"].sum()

    # Create a method that calculates average sales.
    def average_sales(self):
        # Return the mean of the Sales column.
        return self.data["Sales"].mean()

    # Create a method that finds the best product.
    def best_product(self):
        # Find the row index containing the maximum sales value.
        best_index = self.data["Sales"].idxmax()
        # Return the product name from the best row.
        return self.data.loc[best_index, "Product"]

# Create sample sales data.
sales_data = pd.DataFrame({
    # Create the Product column.
    "Product": ["Bag", "Jewelry", "Keychain", "Purse"],
    # Create the Sales column.
    "Sales": [45000, 62000, 28000, 71000]
})

# Create a SalesAnalyzer object.
sales_analyzer = SalesAnalyzer(sales_data)

# Display the total sales.
print("Total Sales:", sales_analyzer.total_sales())

# Display the average sales.
print("Average Sales:", sales_analyzer.average_sales())

# Display the best-selling product.
print("Best Product:", sales_analyzer.best_product())
```

# 16. OOP in Data Science Solved ML Pipeline

This project organizes data preparation, model training and evaluation inside one reusable class.

```python
# Import the required scikit-learn tools.
from sklearn.datasets import load_iris

# Import the train-test split function.
from sklearn.model_selection import train_test_split

# Import the StandardScaler class.
from sklearn.preprocessing import StandardScaler

# Import the LogisticRegression model.
from sklearn.linear_model import LogisticRegression

# Import the accuracy score function.
from sklearn.metrics import accuracy_score

# Create a class named ClassificationPipeline.
class ClassificationPipeline:
    # Define a constructor that receives features and labels.
    def __init__(self, X, y):
        # Store the feature data.
        self.X = X
        # Store the target labels.
        self.y = y
        # Create a StandardScaler object.
        self.scaler = StandardScaler()
        # Create a LogisticRegression model object.
        self.model = LogisticRegression(max_iter=300)

    # Create a method for splitting and scaling the data.
    def prepare_data(self):
        # Split the data into training and testing sets.
        self.X_train, self.X_test, self.y_train, self.y_test = train_test_split(
            self.X,
            self.y,
            test_size=0.2,
            random_state=42
        )
        # Fit the scaler and transform the training data.
        self.X_train = self.scaler.fit_transform(self.X_train)
        # Transform the testing data using the fitted scaler.
        self.X_test = self.scaler.transform(self.X_test)

    # Create a method for training the model.
    def train(self):
        # Train the model using the prepared training data.
        self.model.fit(self.X_train, self.y_train)

    # Create a method for evaluating the model.
    def evaluate(self):
        # Generate predictions for the testing data.
        predictions = self.model.predict(self.X_test)
        # Calculate and return the model accuracy.
        return accuracy_score(self.y_test, predictions)

    # Create a method for predicting new samples.
    def predict(self, new_data):
        # Scale the new data using the fitted scaler.
        scaled_data = self.scaler.transform(new_data)
        # Return the model predictions.
        return self.model.predict(scaled_data)

# Load the Iris dataset.
iris = load_iris()

# Store the feature values.
X = iris.data

# Store the target labels.
y = iris.target

# Create a ClassificationPipeline object.
ml_pipeline = ClassificationPipeline(X, y)

# Prepare the dataset.
ml_pipeline.prepare_data()

# Train the classification model.
ml_pipeline.train()

# Display the model accuracy.
print("Model Accuracy:", ml_pipeline.evaluate())

# Create one new flower sample.
new_flower = [[5.1, 3.5, 1.4, 0.2]]

# Predict the class of the new flower.
print("Predicted Class:", ml_pipeline.predict(new_flower))
```

# 17. OOP in AI Simple Intelligent Agent

An AI agent can be represented as a class with state and decision-making behavior.

```python
# Create a class named SmartAgent.
class SmartAgent:
    # Define a constructor for the agent.
    def __init__(self, name):
        # Store the agent name.
        self.name = name
        # Store the initial energy level.
        self.energy = 100

    # Create a method for making a decision.
    def decide(self, temperature):
        # Check whether the temperature is high.
        if temperature > 35:
            # Return the cooling decision.
            return "Turn on cooling system"
        # Check whether the temperature is low.
        if temperature < 15:
            # Return the heating decision.
            return "Turn on heating system"
        # Return the normal condition decision.
        return "Keep the system unchanged"

    # Create a method for using energy.
    def use_energy(self, amount):
        # Reduce the agent energy level.
        self.energy -= amount

# Create a SmartAgent object.
agent = SmartAgent("Temperature Agent")

# Display the agent decision.
print(agent.decide(40))

# Reduce the agent energy.
agent.use_energy(10)

# Display the remaining energy.
print("Remaining Energy:", agent.energy)
```
